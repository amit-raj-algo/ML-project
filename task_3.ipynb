```python
# =====================================================================
# EXPLORATORY DATA ANALYSIS (EDA) PROJECT
# Dataset: Titanic Passenger Dataset
# =====================================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("=" * 75)
print("          EXPLORATORY DATA ANALYSIS (EDA) PROJECT")
print("=" * 75)


# ============================================================
# 2. LOAD DATASET FROM ONLINE CSV
# ============================================================

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

print("\n✅ Dataset loaded successfully!")

print("Dataset Shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 3. BASIC DATA INFORMATION
# ============================================================

print("\n" + "=" * 75)
print("1. BASIC DATA INFORMATION")
print("=" * 75)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
display(df.dtypes.to_frame(name="Data Type"))

print("\nDataset Information:")
df.info()


# ============================================================
# 4. STATISTICAL SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("2. STATISTICAL SUMMARY")
print("=" * 75)

display(df.describe(include="all").T)


# ============================================================
# 5. MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 75)
print("3. MISSING VALUE ANALYSIS")
print("=" * 75)

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum().values,
    "Missing Percentage": (
        df.isnull().sum().values / len(df) * 100
    ).round(2)
})

missing_report = missing_report.sort_values(
    by="Missing Values",
    ascending=False
)

display(missing_report)


# Missing values visualization

plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False
)

plt.title("Missing Values Heatmap")
plt.xlabel("Columns")
plt.ylabel("Rows")

plt.show()


# ============================================================
# 6. DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 75)
print("4. DUPLICATE ANALYSIS")
print("=" * 75)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)


# ============================================================
# 7. TARGET / SURVIVAL DISTRIBUTION
# ============================================================

print("\n" + "=" * 75)
print("5. SURVIVAL DISTRIBUTION")
print("=" * 75)

print(
    df["survived"]
    .value_counts()
)

print("\nSurvival Percentage:")

survival_percentage = (
    df["survived"]
    .value_counts(normalize=True)
    * 100
).round(2)

print(survival_percentage)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="survived"
)

plt.title("Survival Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()


# ============================================================
# 8. GENDER DISTRIBUTION
# ============================================================

print("\n" + "=" * 75)
print("6. GENDER ANALYSIS")
print("=" * 75)

print(df["sex"].value_counts())

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="sex"
)

plt.title("Passenger Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()


# ============================================================
# 9. PASSENGER CLASS DISTRIBUTION
# ============================================================

print("\n" + "=" * 75)
print("7. PASSENGER CLASS ANALYSIS")
print("=" * 75)

print(df["pclass"].value_counts().sort_index())

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="pclass"
)

plt.title("Passenger Distribution by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()


# ============================================================
# 10. AGE DISTRIBUTION
# ============================================================

print("\n" + "=" * 75)
print("8. AGE DISTRIBUTION")
print("=" * 75)

print("Average Age:", round(df["age"].mean(), 2))
print("Median Age:", round(df["age"].median(), 2))
print("Minimum Age:", df["age"].min())
print("Maximum Age:", df["age"].max())


plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="age",
    bins=30,
    kde=True
)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()


# ============================================================
# 11. FARE DISTRIBUTION
# ============================================================

print("\n" + "=" * 75)
print("9. FARE DISTRIBUTION")
print("=" * 75)

print("Average Fare:", round(df["fare"].mean(), 2))
print("Median Fare:", round(df["fare"].median(), 2))
print("Minimum Fare:", df["fare"].min())
print("Maximum Fare:", df["fare"].max())


plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="fare",
    bins=30,
    kde=True
)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.show()


# ============================================================
# 12. FARE BOXPLOT
# ============================================================

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="fare"
)

plt.title("Fare Boxplot")
plt.xlabel("Fare")

plt.show()


# ============================================================
# 13. AGE BOXPLOT
# ============================================================

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="age"
)

plt.title("Age Boxplot")
plt.xlabel("Age")

plt.show()


# ============================================================
# 14. SURVIVAL BY GENDER
# ============================================================

print("\n" + "=" * 75)
print("10. SURVIVAL BY GENDER")
print("=" * 75)

gender_survival = (
    df.groupby("sex")["survived"]
    .mean()
    .mul(100)
    .round(2)
)

print("Survival Rate by Gender:")
print(gender_survival)


plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="sex",
    y="survived"
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()


# ============================================================
# 15. SURVIVAL BY PASSENGER CLASS
# ============================================================

print("\n" + "=" * 75)
print("11. SURVIVAL BY PASSENGER CLASS")
print("=" * 75)

class_survival = (
    df.groupby("pclass")["survived"]
    .mean()
    .mul(100)
    .round(2)
)

print("Survival Rate by Passenger Class:")
print(class_survival)


plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="pclass",
    y="survived"
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()


# ============================================================
# 16. SURVIVAL BY GENDER AND CLASS
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="pclass",
    y="survived",
    hue="sex"
)

plt.title("Survival Rate by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()


# ============================================================
# 17. SURVIVAL BY EMBARKATION PORT
# ============================================================

print("\n" + "=" * 75)
print("12. SURVIVAL BY EMBARKATION PORT")
print("=" * 75)

print(df["embarked"].value_counts())

embarked_survival = (
    df.groupby("embarked")["survived"]
    .mean()
    .mul(100)
    .round(2)
)

print("\nSurvival Rate by Port:")
print(embarked_survival)


plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="embarked",
    y="survived"
)

plt.title("Survival Rate by Embarkation Port")
plt.xlabel("Embarkation Port")
plt.ylabel("Survival Rate")

plt.show()


# ============================================================
# 18. AGE VS SURVIVAL
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="survived",
    y="age"
)

plt.title("Age Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")

plt.show()


# ============================================================
# 19. FARE VS CLASS
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="pclass",
    y="fare"
)

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

plt.show()


# ============================================================
# 20. AGE VS FARE SCATTER PLOT
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived",
    alpha=0.7
)

plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()


# ============================================================
# 21. FAMILY SIZE ANALYSIS
# ============================================================

df["family_size"] = (
    df["sibsp"] +
    df["parch"] +
    1
)

print("\n" + "=" * 75)
print("13. FAMILY SIZE ANALYSIS")
print("=" * 75)

display(
    df[
        [
            "family_size",
            "survived"
        ]
    ].groupby("family_size")
    .agg(
        passengers=("survived", "count"),
        survival_rate=("survived", "mean")
    )
    .assign(
        survival_rate=lambda x:
        x["survival_rate"] * 100
    )
    .round(2)
)


plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x="family_size",
    y="survived"
)

plt.title("Survival Rate by Family Size")
plt.xlabel("Family Size")
plt.ylabel("Survival Rate")

plt.show()


# ============================================================
# 22. ALONE VS WITH FAMILY
# ============================================================

df["is_alone"] = (
    df["family_size"] == 1
)

df["travel_status"] = np.where(
    df["is_alone"],
    "Alone",
    "With Family"
)

print("\n" + "=" * 75)
print("14. SOLO VS FAMILY TRAVEL")
print("=" * 75)

alone_analysis = (
    df.groupby("travel_status")["survived"]
    .agg(
        passengers="count",
        survival_rate="mean"
    )
)

alone_analysis["survival_rate"] = (
    alone_analysis["survival_rate"] * 100
).round(2)

display(alone_analysis)


plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="travel_status",
    y="survived"
)

plt.title("Survival Rate: Alone vs With Family")
plt.xlabel("Travel Status")
plt.ylabel("Survival Rate")

plt.show()


# ============================================================
# 23. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 75)
print("15. CORRELATION ANALYSIS")
print("=" * 75)

numeric_columns = [
    "survived",
    "pclass",
    "age",
    "sibsp",
    "parch",
    "fare",
    "family_size"
]

correlation_matrix = (
    df[numeric_columns]
    .corr()
)

display(
    correlation_matrix.round(2)
)


plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.show()


# ============================================================
# 24. PAIRPLOT
# ============================================================

print("\n" + "=" * 75)
print("16. PAIRWISE RELATIONSHIPS")
print("=" * 75)

pairplot_data = df[
    [
        "survived",
        "pclass",
        "age",
        "fare",
        "family_size"
    ]
].dropna()

sns.pairplot(
    pairplot_data,
    hue="survived",
    diag_kind="hist"
)

plt.show()


# ============================================================
# 25. AGE GROUP ANALYSIS
# ============================================================

df["age_group"] = pd.cut(
    df["age"],
    bins=[
        0,
        12,
        18,
        30,
        45,
        60,
        100
    ],
    labels=[
        "Child",
        "Teen",
        "Young Adult",
        "Adult",
        "Middle Age",
        "Senior"
    ]
)

age_group_analysis = (
    df.groupby(
        "age_group",
        observed=True
    )["survived"]
    .agg(
        passengers="count",
        survival_rate="mean"
    )
)

age_group_analysis["survival_rate"] = (
    age_group_analysis["survival_rate"] * 100
).round(2)

print("\n" + "=" * 75)
print("17. SURVIVAL BY AGE GROUP")
print("=" * 75)

display(age_group_analysis)


plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="age_group",
    y="survived"
)

plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")

plt.xticks(rotation=30)

plt.show()


# ============================================================
# 26. TOP CORRELATIONS WITH SURVIVAL
# ============================================================

print("\n" + "=" * 75)
print("18. FEATURES MOST RELATED TO SURVIVAL")
print("=" * 75)

survival_corr = (
    correlation_matrix["survived"]
    .drop("survived")
    .sort_values(
        key=abs,
        ascending=False
    )
)

print(
    survival_corr.round(3)
)


# ============================================================
# 27. AUTOMATIC KEY FINDINGS
# ============================================================

print("\n" + "=" * 75)
print("19. KEY FINDINGS")
print("=" * 75)

overall_survival = (
    df["survived"].mean() * 100
)

female_rate = (
    df[df["sex"] == "female"]["survived"].mean()
    * 100
)

male_rate = (
    df[df["sex"] == "male"]["survived"].mean()
    * 100
)

first_class_rate = (
    df[df["pclass"] == 1]["survived"].mean()
    * 100
)

second_class_rate = (
    df[df["pclass"] == 2]["survived"].mean()
    * 100
)

third_class_rate = (
    df[df["pclass"] == 3]["survived"].mean()
    * 100
)

print(
    f"\n1. Overall survival rate: "
    f"{overall_survival:.2f}%"
)

print(
    f"\n2. Female survival rate: "
    f"{female_rate:.2f}%"
)

print(
    f"   Male survival rate: "
    f"{male_rate:.2f}%"
)

print(
    f"\n3. First-class survival rate: "
    f"{first_class_rate:.2f}%"
)

print(
    f"   Second-class survival rate: "
    f"{second_class_rate:.2f}%"
)

print(
    f"   Third-class survival rate: "
    f"{third_class_rate:.2f}%"
)

print(
    f"\n4. Average age: "
    f"{df['age'].mean():.2f} years"
)

print(
    f"\n5. Average fare: "
    f"${df['fare'].mean():.2f}"
)

print(
    f"\n6. Most common passenger class: "
    f"{df['pclass'].mode()[0]}"
)

print(
    f"\n7. Most common embarkation port: "
    f"{df['embarked'].mode()[0]}"
)


# ============================================================
# 28. FINAL DATASET AFTER FEATURE ENGINEERING
# ============================================================

print("\n" + "=" * 75)
print("20. FINAL DATASET")
print("=" * 75)

print(
    "Final Shape:",
    df.shape
)

display(df.head())


# ============================================================
# 29. SAVE ANALYZED DATASET
# ============================================================

df.to_csv(
    "titanic_eda_processed.csv",
    index=False
)

print(
    "\n✅ Processed dataset saved as: "
    "titanic_eda_processed.csv"
)


# ============================================================
# 30. FINAL PROJECT SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("             EDA PROJECT COMPLETED")
print("=" * 75)

print("\nDataset:")
print("Titanic Passenger Dataset")

print("\nEDA Tasks Completed:")
print("✅ Dataset loading")
print("✅ Dataset structure analysis")
print("✅ Data type analysis")
print("✅ Statistical summaries")
print("✅ Missing value analysis")
print("✅ Duplicate analysis")
print("✅ Survival distribution")
print("✅ Gender analysis")
print("✅ Passenger class analysis")
print("✅ Age distribution")
print("✅ Fare distribution")
print("✅ Survival by gender")
print("✅ Survival by passenger class")
print("✅ Survival by embarkation")
print("✅ Age vs survival")
print("✅ Fare vs class")
print("✅ Age vs fare")
print("✅ Family size analysis")
print("✅ Solo vs family analysis")
print("✅ Correlation analysis")
print("✅ Pairplot analysis")
print("✅ Age group analysis")
print("✅ Key influencing factors")
print("✅ Automated key findings")
print("✅ Processed CSV export")

print("\n" + "=" * 75)
print("          ALL EDA REQUIREMENTS COMPLETED!")
print("=" * 75)
```
