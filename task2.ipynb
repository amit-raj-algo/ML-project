```python
# =====================================================================
# PREDICTIVE MODELING USING MACHINE LEARNING
# Dataset: Breast Cancer Wisconsin (Diagnostic)
# Models: Logistic Regression, Decision Tree, Random Forest
# =====================================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier, plot_tree

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)

import joblib

from IPython.display import display

sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)


print("=" * 75)
print("       PREDICTIVE MODELING USING MACHINE LEARNING")
print("=" * 75)


# =====================================================================
# 2. LOAD DATASET ONLINE
# =====================================================================

print("\n" + "=" * 75)
print("1. LOADING DATASET")
print("=" * 75)

# Load official Breast Cancer Wisconsin dataset
data = load_breast_cancer(as_frame=True)

X = data.data.copy()
y = data.target.copy()

# Convert target into meaningful labels
# Original sklearn target:
# 0 = malignant
# 1 = benign

df = X.copy()

df["target"] = y

df["diagnosis"] = df["target"].map({
    0: "Malignant",
    1: "Benign"
})

print("\nDataset loaded successfully!")

print("Number of rows:", df.shape[0])
print("Number of features:", X.shape[1])

print("\nFirst 5 rows:")
display(df.head())


# =====================================================================
# 3. BASIC DATA EXPLORATION
# =====================================================================

print("\n" + "=" * 75)
print("2. BASIC DATA EXPLORATION")
print("=" * 75)

print("\nDataset Shape:")
print(df.shape)

print("\nFeature Names:")
print(X.columns.tolist())

print("\nData Types:")
display(X.dtypes.to_frame(name="Data Type"))

print("\nStatistical Summary:")
display(X.describe().T)


# =====================================================================
# 4. CHECK MISSING VALUES
# =====================================================================

print("\n" + "=" * 75)
print("3. MISSING VALUE ANALYSIS")
print("=" * 75)

missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": missing_values.values,
    "Missing Percentage": (
        missing_values.values / len(df) * 100
    ).round(2)
})

display(
    missing_report[
        missing_report["Missing Values"] > 0
    ]
)

if df.isnull().sum().sum() == 0:
    print("✅ No missing values found.")
else:
    print("⚠ Missing values found.")


# =====================================================================
# 5. CHECK DUPLICATES
# =====================================================================

print("\n" + "=" * 75)
print("4. DUPLICATE ANALYSIS")
print("=" * 75)

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)

if duplicate_count > 0:

    df = df.drop_duplicates().reset_index(drop=True)

    X = df.drop(
        columns=["target", "diagnosis"]
    )

    y = df["target"]

    print("Duplicates removed.")

else:

    print("✅ No duplicate rows found.")


# =====================================================================
# 6. TARGET DISTRIBUTION
# =====================================================================

print("\n" + "=" * 75)
print("5. TARGET DISTRIBUTION")
print("=" * 75)

target_counts = df["diagnosis"].value_counts()

print(target_counts)

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="diagnosis"
)

plt.title("Target Class Distribution")
plt.xlabel("Diagnosis")
plt.ylabel("Number of Samples")

plt.show()


# =====================================================================
# 7. PREPARE FEATURES AND TARGET
# =====================================================================

X = df.drop(
    columns=["target", "diagnosis"]
)

y = df["target"]

print("\nFeatures shape:", X.shape)
print("Target shape:", y.shape)


# =====================================================================
# 8. TRAIN TEST SPLIT
# =====================================================================

print("\n" + "=" * 75)
print("6. TRAIN / TEST SPLIT")
print("=" * 75)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])


# =====================================================================
# 9. FEATURE SCALING
# =====================================================================

print("\n" + "=" * 75)
print("7. FEATURE SCALING")
print("=" * 75)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("✅ StandardScaler applied.")
print("Training data scaled successfully.")


# =====================================================================
# 10. LOGISTIC REGRESSION
# =====================================================================

print("\n" + "=" * 75)
print("8. LOGISTIC REGRESSION")
print("=" * 75)

logistic_model = LogisticRegression(
    max_iter=5000,
    random_state=42
)

logistic_model.fit(
    X_train_scaled,
    y_train
)

logistic_pred = logistic_model.predict(
    X_test_scaled
)

logistic_prob = logistic_model.predict_proba(
    X_test_scaled
)[:, 1]

logistic_accuracy = accuracy_score(
    y_test,
    logistic_pred
)

logistic_precision = precision_score(
    y_test,
    logistic_pred
)

logistic_recall = recall_score(
    y_test,
    logistic_pred
)

logistic_f1 = f1_score(
    y_test,
    logistic_pred
)

logistic_auc = roc_auc_score(
    y_test,
    logistic_prob
)

print(f"Accuracy : {logistic_accuracy:.4f}")
print(f"Precision: {logistic_precision:.4f}")
print(f"Recall   : {logistic_recall:.4f}")
print(f"F1 Score : {logistic_f1:.4f}")
print(f"ROC AUC  : {logistic_auc:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        logistic_pred,
        target_names=[
            "Malignant",
            "Benign"
        ]
    )
)


# =====================================================================
# 11. DECISION TREE
# =====================================================================

print("\n" + "=" * 75)
print("9. DECISION TREE")
print("=" * 75)

tree_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

tree_model.fit(
    X_train,
    y_train
)

tree_pred = tree_model.predict(
    X_test
)

tree_prob = tree_model.predict_proba(
    X_test
)[:, 1]

tree_accuracy = accuracy_score(
    y_test,
    tree_pred
)

tree_precision = precision_score(
    y_test,
    tree_pred
)

tree_recall = recall_score(
    y_test,
    tree_pred
)

tree_f1 = f1_score(
    y_test,
    tree_pred
)

tree_auc = roc_auc_score(
    y_test,
    tree_prob
)

print(f"Accuracy : {tree_accuracy:.4f}")
print(f"Precision: {tree_precision:.4f}")
print(f"Recall   : {tree_recall:.4f}")
print(f"F1 Score : {tree_f1:.4f}")
print(f"ROC AUC  : {tree_auc:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        tree_pred,
        target_names=[
            "Malignant",
            "Benign"
        ]
    )
)


# =====================================================================
# 12. RANDOM FOREST
# =====================================================================

print("\n" + "=" * 75)
print("10. RANDOM FOREST")
print("=" * 75)

random_forest_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    random_state=42
)

random_forest_model.fit(
    X_train,
    y_train
)

rf_pred = random_forest_model.predict(
    X_test
)

rf_prob = random_forest_model.predict_proba(
    X_test
)[:, 1]

rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

rf_precision = precision_score(
    y_test,
    rf_pred
)

rf_recall = recall_score(
    y_test,
    rf_pred
)

rf_f1 = f1_score(
    y_test,
    rf_pred
)

rf_auc = roc_auc_score(
    y_test,
    rf_prob
)

print(f"Accuracy : {rf_accuracy:.4f}")
print(f"Precision: {rf_precision:.4f}")
print(f"Recall   : {rf_recall:.4f}")
print(f"F1 Score : {rf_f1:.4f}")
print(f"ROC AUC  : {rf_auc:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        rf_pred,
        target_names=[
            "Malignant",
            "Benign"
        ]
    )
)


# =====================================================================
# 13. MODEL COMPARISON
# =====================================================================

print("\n" + "=" * 75)
print("11. MODEL COMPARISON")
print("=" * 75)

results = pd.DataFrame({

    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "Accuracy": [
        logistic_accuracy,
        tree_accuracy,
        rf_accuracy
    ],

    "Precision": [
        logistic_precision,
        tree_precision,
        rf_precision
    ],

    "Recall": [
        logistic_recall,
        tree_recall,
        rf_recall
    ],

    "F1 Score": [
        logistic_f1,
        tree_f1,
        rf_f1
    ],

    "ROC AUC": [
        logistic_auc,
        tree_auc,
        rf_auc
    ]
})

results = results.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

display(
    results.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}",
        "ROC AUC": "{:.4f}"
    })
)


# =====================================================================
# 14. MODEL ACCURACY VISUALIZATION
# =====================================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    data=results,
    x="Model",
    y="Accuracy"
)

plt.title("Model Accuracy Comparison")
plt.xlabel("Machine Learning Model")
plt.ylabel("Accuracy")

plt.ylim(
    max(0, results["Accuracy"].min() - 0.05),
    1.0
)

plt.xticks(rotation=15)

plt.show()


# =====================================================================
# 15. CONFUSION MATRIX - LOGISTIC REGRESSION
# =====================================================================

cm_logistic = confusion_matrix(
    y_test,
    logistic_pred
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_logistic,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Malignant",
        "Benign"
    ],
    yticklabels=[
        "Malignant",
        "Benign"
    ]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()


# =====================================================================
# 16. CONFUSION MATRIX - DECISION TREE
# =====================================================================

cm_tree = confusion_matrix(
    y_test,
    tree_pred
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_tree,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=[
        "Malignant",
        "Benign"
    ],
    yticklabels=[
        "Malignant",
        "Benign"
    ]
)

plt.title("Confusion Matrix - Decision Tree")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()


# =====================================================================
# 17. CONFUSION MATRIX - RANDOM FOREST
# =====================================================================

cm_rf = confusion_matrix(
    y_test,
    rf_pred
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="Oranges",
    xticklabels=[
        "Malignant",
        "Benign"
    ],
    yticklabels=[
        "Malignant",
        "Benign"
    ]
)

plt.title("Confusion Matrix - Random Forest")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()


# =====================================================================
# 18. ROC CURVES
# =====================================================================

logistic_fpr, logistic_tpr, _ = roc_curve(
    y_test,
    logistic_prob
)

tree_fpr, tree_tpr, _ = roc_curve(
    y_test,
    tree_prob
)

rf_fpr, rf_tpr, _ = roc_curve(
    y_test,
    rf_prob
)

plt.figure(figsize=(10, 7))

plt.plot(
    logistic_fpr,
    logistic_tpr,
    label=f"Logistic Regression (AUC = {logistic_auc:.3f})"
)

plt.plot(
    tree_fpr,
    tree_tpr,
    label=f"Decision Tree (AUC = {tree_auc:.3f})"
)

plt.plot(
    rf_fpr,
    rf_tpr,
    label=f"Random Forest (AUC = {rf_auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Guess"
)

plt.title("ROC Curves - Model Comparison")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.legend()

plt.show()


# =====================================================================
# 19. RANDOM FOREST FEATURE IMPORTANCE
# =====================================================================

print("\n" + "=" * 75)
print("12. RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 75)

feature_importance = pd.DataFrame({

    "Feature": X.columns,

    "Importance": (
        random_forest_model.feature_importances_
    )

})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

display(
    feature_importance.head(15)
)


# Plot top 15 features

top_features = feature_importance.head(15)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=top_features,
    x="Importance",
    y="Feature"
)

plt.title(
    "Top 15 Important Features - Random Forest"
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.show()


# =====================================================================
# 20. DECISION TREE VISUALIZATION
# =====================================================================

plt.figure(figsize=(22, 12))

plot_tree(
    tree_model,
    feature_names=X.columns,
    class_names=[
        "Malignant",
        "Benign"
    ],
    filled=True,
    max_depth=3,
    fontsize=8
)

plt.title("Decision Tree Visualization")
plt.show()


# =====================================================================
# 21. BEST MODEL
# =====================================================================

best_model_name = results.iloc[0]["Model"]
best_accuracy = results.iloc[0]["Accuracy"]

print("\n" + "=" * 75)
print("13. BEST MODEL")
print("=" * 75)

print(
    f"Best Model: {best_model_name}"
)

print(
    f"Best Accuracy: {best_accuracy:.4f}"
)

print(
    f"Best Accuracy Percentage: "
    f"{best_accuracy * 100:.2f}%"
)


# =====================================================================
# 22. SAVE CLEAN DATASET
# =====================================================================

clean_dataset_file = (
    "breast_cancer_cleaned_dataset.csv"
)

df.to_csv(
    clean_dataset_file,
    index=False
)

print(
    f"\n✅ Clean dataset saved as: "
    f"{clean_dataset_file}"
)


# =====================================================================
# 23. SAVE MODELS
# =====================================================================

joblib.dump(
    logistic_model,
    "logistic_regression_model.pkl"
)

joblib.dump(
    tree_model,
    "decision_tree_model.pkl"
)

joblib.dump(
    random_forest_model,
    "random_forest_model.pkl"
)

joblib.dump(
    scaler,
    "feature_scaler.pkl"
)

print("✅ Models saved successfully.")


# =====================================================================
# 24. FINAL PROJECT SUMMARY
# =====================================================================

print("\n" + "=" * 75)
print("             PROJECT COMPLETED SUCCESSFULLY")
print("=" * 75)

print("\nDataset:")
print("Breast Cancer Wisconsin (Diagnostic)")

print("\nDataset Details:")
print(f"Total Samples: {len(df)}")
print(f"Total Features: {X.shape[1]}")

print("\nMachine Learning Models:")
print("✅ Logistic Regression")
print("✅ Decision Tree")
print("✅ Random Forest")

print("\nModel Evaluation:")
print("✅ Accuracy")
print("✅ Precision")
print("✅ Recall")
print("✅ F1 Score")
print("✅ Confusion Matrix")
print("✅ ROC Curve")
print("✅ ROC AUC")

print("\nVisualizations:")
print("✅ Target Distribution")
print("✅ Model Accuracy Comparison")
print("✅ 3 Confusion Matrices")
print("✅ ROC Curve Comparison")
print("✅ Random Forest Feature Importance")
print("✅ Decision Tree Visualization")

print("\nGenerated Files:")
print("📁 breast_cancer_cleaned_dataset.csv")
print("📁 logistic_regression_model.pkl")
print("📁 decision_tree_model.pkl")
print("📁 random_forest_model.pkl")
print("📁 feature_scaler.pkl")

print("\n" + "=" * 75)
print("       PREDICTIVE MODELING TASK COMPLETED!")
print("=" * 75)
```
