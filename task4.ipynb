```python
# =====================================================================
# REAL-WORLD DATA PROJECT
# Domain: RETAIL / SALES ANALYTICS
# Dataset: Superstore Sales Dataset
# =====================================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("=" * 80)
print("              REAL-WORLD DATA PROJECT")
print("                 RETAIL SALES ANALYSIS")
print("=" * 80)


# ============================================================
# 2. LOAD DATASET FROM ONLINE CSV
# ============================================================

url = "https://raw.githubusercontent.com/leonism/sample-superstore/master/data/superstore.csv"

df = pd.read_csv(url)

print("\n✅ Dataset loaded successfully!")
print("Original Dataset Shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 3. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

print("\nCleaned Columns:")
print(df.columns.tolist())


# ============================================================
# 4. BASIC DATA EXPLORATION
# ============================================================

print("\n" + "=" * 80)
print("1. DATASET OVERVIEW")
print("=" * 80)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nData Types:")
display(df.dtypes.to_frame(name="Data Type"))

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all").T)


# ============================================================
# 5. CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 80)
print("2. DATA QUALITY CHECK")
print("=" * 80)

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum().values,
    "Missing %": (
        df.isnull().sum().values /
        len(df) * 100
    ).round(2),
    "Unique Values": [
        df[col].nunique()
        for col in df.columns
    ]
})

missing_report = missing_report.sort_values(
    "Missing Values",
    ascending=False
)

display(missing_report)


# ============================================================
# 6. DUPLICATE CHECK
# ============================================================

duplicate_count = df.duplicated().sum()

print("\nDuplicate Rows:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Duplicates removed.")
else:
    print("✅ No duplicate rows found.")


# ============================================================
# 7. CONVERT DATE COLUMNS
# ============================================================

# Detect order date column
if "order_date" in df.columns:
    df["order_date"] = pd.to_datetime(
        df["order_date"],
        errors="coerce"
    )

# Detect ship date column
if "ship_date" in df.columns:
    df["ship_date"] = pd.to_datetime(
        df["ship_date"],
        errors="coerce"
    )

print("\nDate columns converted successfully.")


# ============================================================
# 8. CREATE BUSINESS FEATURES
# ============================================================

# Profit margin
df["profit_margin"] = np.where(
    df["sales"] != 0,
    (df["profit"] / df["sales"]) * 100,
    0
)

# Order year
df["order_year"] = df["order_date"].dt.year

# Order month
df["order_month"] = df["order_date"].dt.month

# Month name
df["month_name"] = df["order_date"].dt.month_name()

# Quarter
df["quarter"] = df["order_date"].dt.quarter

# Shipping days
if "ship_date" in df.columns:
    df["shipping_days"] = (
        df["ship_date"] -
        df["order_date"]
    ).dt.days

print("\n✅ Business features created:")
print("• Profit Margin")
print("• Order Year")
print("• Order Month")
print("• Month Name")
print("• Quarter")
print("• Shipping Days")


# ============================================================
# 9. FINAL MISSING VALUE CHECK
# ============================================================

print("\nMissing Values After Processing:")

remaining_missing = df.isnull().sum()

display(
    remaining_missing[
        remaining_missing > 0
    ].to_frame(name="Missing Values")
)


# ============================================================
# 10. BUSINESS KPI SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("3. BUSINESS KPI SUMMARY")
print("=" * 80)

total_sales = df["sales"].sum()
total_profit = df["profit"].sum()
total_quantity = df["quantity"].sum()
average_order_sales = df["sales"].mean()
average_profit = df["profit"].mean()
average_discount = df["discount"].mean() * 100
overall_profit_margin = (
    total_profit / total_sales * 100
)

unique_orders = df["order_id"].nunique()
unique_customers = df["customer_id"].nunique()
unique_products = df["product_id"].nunique()

print(f"\n💰 Total Sales       : ${total_sales:,.2f}")
print(f"💵 Total Profit      : ${total_profit:,.2f}")
print(f"📦 Total Quantity    : {total_quantity:,}")
print(f"🧾 Total Orders      : {unique_orders:,}")
print(f"👥 Unique Customers  : {unique_customers:,}")
print(f"🛍️ Unique Products   : {unique_products:,}")
print(f"📊 Avg Order Sales   : ${average_order_sales:,.2f}")
print(f"📈 Avg Profit/Row    : ${average_profit:,.2f}")
print(f"🏷️ Avg Discount      : {average_discount:.2f}%")
print(f"📌 Profit Margin     : {overall_profit_margin:.2f}%")


# ============================================================
# 11. SALES BY CATEGORY
# ============================================================

print("\n" + "=" * 80)
print("4. SALES & PROFIT BY CATEGORY")
print("=" * 80)

category_analysis = (
    df.groupby("category")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum"),
        Quantity=("quantity", "sum"),
        Orders=("order_id", "nunique")
    )
    .sort_values(
        "Sales",
        ascending=False
    )
)

category_analysis["Profit Margin %"] = (
    category_analysis["Profit"] /
    category_analysis["Sales"] * 100
)

display(
    category_analysis.round(2)
)


# Sales chart

plt.figure(figsize=(9, 5))

sns.barplot(
    data=category_analysis.reset_index(),
    x="category",
    y="Sales"
)

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()


# Profit chart

plt.figure(figsize=(9, 5))

sns.barplot(
    data=category_analysis.reset_index(),
    x="category",
    y="Profit"
)

plt.title("Profit by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.show()


# ============================================================
# 12. SALES BY SUB-CATEGORY
# ============================================================

print("\n" + "=" * 80)
print("5. TOP SUB-CATEGORIES")
print("=" * 80)

subcategory_analysis = (
    df.groupby("sub_category")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum"),
        Quantity=("quantity", "sum")
    )
    .sort_values(
        "Sales",
        ascending=False
    )
)

display(
    subcategory_analysis.head(15).round(2)
)


# Top subcategories by sales

plt.figure(figsize=(11, 7))

top_subcategories = (
    subcategory_analysis
    .head(10)
    .sort_values("Sales")
)

sns.barplot(
    data=top_subcategories.reset_index(),
    x="Sales",
    y="sub_category"
)

plt.title("Top 10 Sub-Categories by Sales")
plt.xlabel("Sales")
plt.ylabel("Sub-Category")

plt.show()


# ============================================================
# 13. REGIONAL ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("6. REGIONAL PERFORMANCE")
print("=" * 80)

region_analysis = (
    df.groupby("region")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum"),
        Quantity=("quantity", "sum"),
        Orders=("order_id", "nunique")
    )
)

region_analysis["Profit Margin %"] = (
    region_analysis["Profit"] /
    region_analysis["Sales"] * 100
)

display(
    region_analysis.round(2)
)


# Regional sales

plt.figure(figsize=(9, 5))

sns.barplot(
    data=region_analysis.reset_index(),
    x="region",
    y="Sales"
)

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()


# Regional profit

plt.figure(figsize=(9, 5))

sns.barplot(
    data=region_analysis.reset_index(),
    x="region",
    y="Profit"
)

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")

plt.show()


# ============================================================
# 14. CUSTOMER SEGMENT ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("7. CUSTOMER SEGMENT ANALYSIS")
print("=" * 80)

segment_analysis = (
    df.groupby("segment")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum"),
        Quantity=("quantity", "sum"),
        Orders=("order_id", "nunique")
    )
)

segment_analysis["Profit Margin %"] = (
    segment_analysis["Profit"] /
    segment_analysis["Sales"] * 100
)

display(
    segment_analysis.round(2)
)


plt.figure(figsize=(9, 5))

sns.barplot(
    data=segment_analysis.reset_index(),
    x="segment",
    y="Sales"
)

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Sales")

plt.show()


# ============================================================
# 15. SALES TREND OVER TIME
# ============================================================

print("\n" + "=" * 80)
print("8. SALES TREND ANALYSIS")
print("=" * 80)

monthly_sales = (
    df.groupby(
        df["order_date"].dt.to_period("M")
    )
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum")
    )
    .reset_index()
)

monthly_sales["order_date"] = (
    monthly_sales["order_date"]
    .dt.to_timestamp()
)

display(
    monthly_sales.head(12)
)


# Monthly sales trend

plt.figure(figsize=(15, 6))

plt.plot(
    monthly_sales["order_date"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# Monthly profit trend

plt.figure(figsize=(15, 6))

plt.plot(
    monthly_sales["order_date"],
    monthly_sales["Profit"],
    marker="o"
)

plt.title("Monthly Profit Trend")
plt.xlabel("Date")
plt.ylabel("Profit")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 16. YEARLY PERFORMANCE
# ============================================================

print("\n" + "=" * 80)
print("9. YEARLY SALES & PROFIT")
print("=" * 80)

yearly_analysis = (
    df.groupby("order_year")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum"),
        Quantity=("quantity", "sum"),
        Orders=("order_id", "nunique")
    )
)

yearly_analysis["Profit Margin %"] = (
    yearly_analysis["Profit"] /
    yearly_analysis["Sales"] * 100
)

display(
    yearly_analysis.round(2)
)


# ============================================================
# 17. DISCOUNT VS PROFIT ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("10. DISCOUNT VS PROFIT ANALYSIS")
print("=" * 80)

discount_analysis = (
    df.groupby("discount")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum"),
        Orders=("order_id", "nunique")
    )
    .reset_index()
)

discount_analysis["Profit Margin %"] = (
    discount_analysis["Profit"] /
    discount_analysis["Sales"] * 100
)

display(
    discount_analysis.round(2)
)


plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df.sample(
        min(3000, len(df)),
        random_state=42
    ),
    x="discount",
    y="profit",
    alpha=0.6
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()


# ============================================================
# 18. SALES VS PROFIT
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df.sample(
        min(3000, len(df)),
        random_state=42
    ),
    x="sales",
    y="profit",
    hue="category",
    alpha=0.7
)

plt.title("Sales vs Profit by Category")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()


# ============================================================
# 19. PROFIT MARGIN BY CATEGORY
# ============================================================

margin_category = (
    df.groupby("category")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum")
    )
)

margin_category["Profit Margin %"] = (
    margin_category["Profit"] /
    margin_category["Sales"] * 100
)

print("\nProfit Margin by Category:")

display(
    margin_category.round(2)
)


plt.figure(figsize=(9, 5))

sns.barplot(
    data=margin_category.reset_index(),
    x="category",
    y="Profit Margin %"
)

plt.title("Profit Margin by Category")
plt.xlabel("Category")
plt.ylabel("Profit Margin (%)")

plt.show()


# ============================================================
# 20. TOP 10 PRODUCTS BY SALES
# ============================================================

print("\n" + "=" * 80)
print("11. TOP 10 PRODUCTS")
print("=" * 80)

top_products = (
    df.groupby("product_name")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum"),
        Quantity=("quantity", "sum")
    )
    .sort_values(
        "Sales",
        ascending=False
    )
    .head(10)
)

display(
    top_products.round(2)
)


# ============================================================
# 21. TOP 10 PRODUCTS BY PROFIT
# ============================================================

print("\nTop 10 Products by Profit:")

top_profit_products = (
    df.groupby("product_name")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum"),
        Quantity=("quantity", "sum")
    )
    .sort_values(
        "Profit",
        ascending=False
    )
    .head(10)
)

display(
    top_profit_products.round(2)
)


# ============================================================
# 22. LOSS-MAKING PRODUCTS
# ============================================================

print("\n" + "=" * 80)
print("12. LOSS-MAKING PRODUCTS")
print("=" * 80)

loss_products = (
    df.groupby("product_name")
    .agg(
        Sales=("sales", "sum"),
        Profit=("profit", "sum")
    )
    .sort_values(
        "Profit",
        ascending=True
    )
    .head(10)
)

display(
    loss_products.round(2)
)


# ============================================================
# 23. SHIPPING ANALYSIS
# ============================================================

if "shipping_days" in df.columns:

    print("\n" + "=" * 80)
    print("13. SHIPPING PERFORMANCE")
    print("=" * 80)

    shipping_analysis = (
        df.groupby("ship_mode")
        .agg(
            Average_Shipping_Days=(
                "shipping_days",
                "mean"
            ),
            Orders=(
                "order_id",
                "nunique"
            )
        )
        .sort_values(
            "Average_Shipping_Days"
        )
    )

    display(
        shipping_analysis.round(2)
    )


    plt.figure(figsize=(10, 5))

    sns.barplot(
        data=shipping_analysis.reset_index(),
        x="ship_mode",
        y="Average_Shipping_Days"
    )

    plt.title("Average Shipping Days by Ship Mode")
    plt.xlabel("Ship Mode")
    plt.ylabel("Average Shipping Days")

    plt.xticks(rotation=20)

    plt.show()


# ============================================================
# 24. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("14. CORRELATION ANALYSIS")
print("=" * 80)

correlation_columns = [
    "sales",
    "quantity",
    "discount",
    "profit",
    "profit_margin"
]

correlation_matrix = (
    df[correlation_columns]
    .corr()
)

display(
    correlation_matrix.round(2)
)


plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Retail Business Correlation Heatmap")

plt.show()


# ============================================================
# 25. REGION × CATEGORY ANALYSIS
# ============================================================

region_category = pd.pivot_table(
    df,
    values="sales",
    index="region",
    columns="category",
    aggfunc="sum"
)

print("\n" + "=" * 80)
print("15. SALES BY REGION AND CATEGORY")
print("=" * 80)

display(
    region_category.round(2)
)


plt.figure(figsize=(10, 6))

sns.heatmap(
    region_category,
    annot=True,
    fmt=".0f",
    cmap="YlGnBu"
)

plt.title("Sales Heatmap: Region vs Category")

plt.show()


# ============================================================
# 26. CUSTOMER SEGMENT × CATEGORY
# ============================================================

segment_category = pd.pivot_table(
    df,
    values="sales",
    index="segment",
    columns="category",
    aggfunc="sum"
)

print("\nSales by Customer Segment and Category:")

display(
    segment_category.round(2)
)


plt.figure(figsize=(10, 6))

sns.heatmap(
    segment_category,
    annot=True,
    fmt=".0f",
    cmap="Oranges"
)

plt.title("Sales: Customer Segment vs Category")

plt.show()


# ============================================================
# 27. BUSINESS DASHBOARD
# ============================================================

print("\n" + "=" * 80)
print("16. EXECUTIVE BUSINESS DASHBOARD")
print("=" * 80)

fig, axes = plt.subplots(
    2,
    2,
    figsize=(16, 11)
)

# Chart 1: Category sales

sns.barplot(
    data=category_analysis.reset_index(),
    x="category",
    y="Sales",
    ax=axes[0, 0]
)

axes[0, 0].set_title(
    "Sales by Category"
)


# Chart 2: Regional profit

sns.barplot(
    data=region_analysis.reset_index(),
    x="region",
    y="Profit",
    ax=axes[0, 1]
)

axes[0, 1].set_title(
    "Profit by Region"
)


# Chart 3: Segment sales

sns.barplot(
    data=segment_analysis.reset_index(),
    x="segment",
    y="Sales",
    ax=axes[1, 0]
)

axes[1, 0].set_title(
    "Sales by Customer Segment"
)


# Chart 4: Monthly sales

axes[1, 1].plot(
    monthly_sales["order_date"],
    monthly_sales["Sales"],
    marker="o"
)

axes[1, 1].set_title(
    "Monthly Sales Trend"
)

axes[1, 1].tick_params(
    axis="x",
    rotation=45
)

plt.suptitle(
    "SUPERSTORE RETAIL BUSINESS DASHBOARD",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()

plt.show()


# ============================================================
# 28. KEY BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 80)
print("17. KEY BUSINESS INSIGHTS")
print("=" * 80)


# Best category

best_category = (
    category_analysis["Sales"]
    .idxmax()
)

best_category_sales = (
    category_analysis.loc[
        best_category,
        "Sales"
    ]
)


# Most profitable category

most_profitable_category = (
    category_analysis["Profit"]
    .idxmax()
)


# Best region

best_region = (
    region_analysis["Sales"]
    .idxmax()
)


# Most profitable region

most_profitable_region = (
    region_analysis["Profit"]
    .idxmax()
)


# Best segment

best_segment = (
    segment_analysis["Sales"]
    .idxmax()
)


# Best product

best_product = (
    top_products.index[0]
)


# Worst product

worst_product = (
    loss_products.index[0]
)


# Highest sales month

highest_sales_month = (
    monthly_sales.loc[
        monthly_sales["Sales"].idxmax()
    ]
)


# Highest profit month

highest_profit_month = (
    monthly_sales.loc[
        monthly_sales["Profit"].idxmax()
    ]
)


print(
    f"\n1. Best sales category: "
    f"{best_category}"
)

print(
    f"   Sales: "
    f"${best_category_sales:,.2f}"
)

print(
    f"\n2. Most profitable category: "
    f"{most_profitable_category}"
)

print(
    f"\n3. Best sales region: "
    f"{best_region}"
)

print(
    f"\n4. Most profitable region: "
    f"{most_profitable_region}"
)

print(
    f"\n5. Best customer segment by sales: "
    f"{best_segment}"
)

print(
    f"\n6. Top product by sales: "
    f"{best_product}"
)

print(
    f"\n7. Lowest-profit product: "
    f"{worst_product}"
)

print(
    f"\n8. Highest-sales month: "
    f"{highest_sales_month['order_date'].strftime('%B %Y')}"
)

print(
    f"   Sales: "
    f"${highest_sales_month['Sales']:,.2f}"
)

print(
    f"\n9. Highest-profit month: "
    f"{highest_profit_month['order_date'].strftime('%B %Y')}"
)

print(
    f"   Profit: "
    f"${highest_profit_month['Profit']:,.2f}"
)


# ============================================================
# 29. AUTOMATIC RECOMMENDATIONS
# ============================================================

print("\n" + "=" * 80)
print("18. BUSINESS RECOMMENDATIONS")
print("=" * 80)

# Discount analysis

high_discount = df[
    df["discount"] >= 0.30
]

if len(high_discount) > 0:

    high_discount_profit = (
        high_discount["profit"].sum()
    )

    print(
        f"\n• High-discount transactions "
        f"(≥30%) generated total profit of "
        f"${high_discount_profit:,.2f}."
    )

print(
    f"\n• Focus marketing efforts on the "
    f"{best_category} category because it "
    f"generates the highest sales."
)

print(
    f"\n• The {best_region} region has the "
    f"highest sales and should be considered "
    f"for continued business expansion."
)

print(
    f"\n• Review loss-making products such as "
    f"'{worst_product}' and evaluate pricing, "
    f"discounts and product costs."
)

print(
    "\n• Monitor discounts carefully because "
    "excessive discounting can reduce profit margins."
)

print(
    "\n• Use monthly sales trends for inventory "
    "planning and seasonal marketing campaigns."
)


# ============================================================
# 30. SAVE PROCESSED DATASET
# ============================================================

output_file = "superstore_real_world_analysis.csv"

df.to_csv(
    output_file,
    index=False
)

print("\n" + "=" * 80)
print("19. OUTPUT")
print("=" * 80)

print(
    f"\n✅ Processed dataset saved as: "
    f"{output_file}"
)


# ============================================================
# 31. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("           REAL-WORLD DATA PROJECT COMPLETED")
print("=" * 80)

print("\nDomain:")
print("Retail / Sales Analytics")

print("\nDataset:")
print("Superstore Sales Dataset")

print("\nAnalysis Completed:")
print("✅ Dataset loading")
print("✅ Data cleaning")
print("✅ Missing value analysis")
print("✅ Duplicate analysis")
print("✅ Date processing")
print("✅ Feature engineering")
print("✅ Sales analysis")
print("✅ Profit analysis")
print("✅ Category analysis")
print("✅ Sub-category analysis")
print("✅ Regional analysis")
print("✅ Customer segment analysis")
print("✅ Time-series analysis")
print("✅ Discount analysis")
print("✅ Product analysis")
print("✅ Shipping analysis")
print("✅ Correlation analysis")
print("✅ Business dashboard")
print("✅ Business insights")
print("✅ Business recommendations")
print("✅ Processed CSV export")

print("\nKey KPIs:")
print(f"Total Sales      : ${total_sales:,.2f}")
print(f"Total Profit     : ${total_profit:,.2f}")
print(f"Total Orders     : {unique_orders:,}")
print(f"Total Customers  : {unique_customers:,}")
print(f"Profit Margin    : {overall_profit_margin:.2f}%")

print("\n" + "=" * 80)
print("             PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 80)
```
