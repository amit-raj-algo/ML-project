```python
# ================================================================
# DATA CLEANING & VISUALIZATION PROJECT
# Dataset: Titanic Passenger Dataset
# ================================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from IPython.display import display
from google.colab import files

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("=" * 70)
print("       DATA CLEANING & VISUALIZATION PROJECT")
print("=" * 70)


# -----------------------------
# 2. LOAD CSV DATASET
# -----------------------------

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("\n✅ Dataset loaded successfully!")
print("Original Dataset Shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())


# -----------------------------
# 3. BASIC DATA EXPLORATION
# -----------------------------

print("\n" + "=" * 70)
print("1. BASIC DATA EXPLORATION")
print("=" * 70)

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
display(df.dtypes.to_frame(name="Data Type"))

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())


# -----------------------------
# 4. MISSING VALUE ANALYSIS
# -----------------------------

print("\n" + "=" * 70)
print("2. MISSING VALUE ANALYSIS")
print("=" * 70)

missing_data = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_data = missing_data.sort_values(
    by="Missing Values",
    ascending=False
)

display(missing_data)

# Missing values heatmap

plt.figure(figsize=(12, 5))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False
)

plt.title("Missing Values Heatmap")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()


# -----------------------------
# 5. DUPLICATE CHECK
# -----------------------------

print("\n" + "=" * 70)
print("3. DUPLICATE ANALYSIS")
print("=" * 70)

duplicates = df.duplicated().sum()

print("Duplicate rows found:", duplicates)

# Remove duplicates
df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)


# -----------------------------
# 6. CLEAN COLUMN NAMES
# -----------------------------

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("\nCleaned Column Names:")
print(df.columns.tolist())


# -----------------------------
# 7. HANDLE MISSING VALUES
# -----------------------------

print("\n" + "=" * 70)
print("4. DATA CLEANING")
print("=" * 70)

# Age → median
df["age"] = df["age"].fillna(
    df["age"].median()
)

# Embarked → mode
df["embarked"] = df["embarked"].fillna(
    df["embarked"].mode()[0]
)

# Cabin has many missing values, so remove it
df = df.drop(columns=["cabin"])

# Convert categorical columns
df["sex"] = df["sex"].astype("category")
df["embarked"] = df["embarked"].astype("category")
df["pclass"] = df["pclass"].astype("category")

print("✅ Missing values handled.")
print("✅ Cabin column removed due to excessive missing values.")
print("✅ Categorical columns converted to category type.")

print("\nMissing values after cleaning:")
display(
    df.isnull().sum().to_frame(
        name="Missing Values"
    )
)


# -----------------------------
# 8. OUTLIER DETECTION
# -----------------------------

print("\n" + "=" * 70)
print("5. OUTLIER DETECTION USING IQR")
print("=" * 70)

numeric_columns = [
    "age",
    "sibsp",
    "parch",
    "fare"
]

outlier_results = []

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_results.append({
        "Column": column,
        "Q1": round(Q1, 2),
        "Q3": round(Q3, 2),
        "IQR": round(IQR, 2),
        "Lower Limit": round(lower_limit, 2),
        "Upper Limit": round(upper_limit, 2),
        "Outliers": len(outliers),
        "Outlier %": round(
            len(outliers) / len(df) * 100,
            2
        )
    })

outlier_df = pd.DataFrame(outlier_results)

display(outlier_df)


# -----------------------------
# 9. OUTLIER VISUALIZATION
# -----------------------------

for column in numeric_columns:

    plt.figure(figsize=(9, 4))

    sns.boxplot(
        x=df[column]
    )

    plt.title(f"Boxplot - {column}")
    plt.xlabel(column)

    plt.show()


# -----------------------------
# 10. HANDLE OUTLIERS
# -----------------------------

df_clean = df.copy()

for column in numeric_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df_clean[column] = df_clean[column].clip(
        lower=lower_limit,
        upper=upper_limit
    )

print("\n✅ Outliers handled using IQR capping.")


# -----------------------------
# 11. FINAL CLEAN DATASET
# -----------------------------

print("\n" + "=" * 70)
print("6. FINAL CLEAN DATASET")
print("=" * 70)

print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])

display(df_clean.head())

print("\nRemaining missing values:")
print(df_clean.isnull().sum().sum())

print("\nRemaining duplicate rows:")
print(df_clean.duplicated().sum())


# ================================================================
# VISUALIZATION SECTION
# ================================================================


# -----------------------------
# 12. SURVIVAL COUNT
# -----------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_clean,
    x="survived"
)

plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()


# -----------------------------
# 13. SURVIVAL BY GENDER
# -----------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_clean,
    x="sex",
    hue="survived"
)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()


# -----------------------------
# 14. SURVIVAL BY CLASS
# -----------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_clean,
    x="pclass",
    hue="survived"
)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()


# -----------------------------
# 15. AGE DISTRIBUTION
# -----------------------------

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df_clean,
    x="age",
    bins=30,
    kde=True
)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()


# -----------------------------
# 16. FARE DISTRIBUTION
# -----------------------------

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df_clean,
    x="fare",
    bins=30,
    kde=True
)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.show()


# -----------------------------
# 17. SURVIVAL RATE BY GENDER
# -----------------------------

gender_survival = (
    df_clean
    .groupby("sex", observed=True)["survived"]
    .mean()
    .reset_index()
)

gender_survival["survival_rate"] = (
    gender_survival["survived"] * 100
)

print("\nSurvival Rate by Gender:")
display(gender_survival)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=gender_survival,
    x="sex",
    y="survival_rate"
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")

plt.show()


# -----------------------------
# 18. SURVIVAL RATE BY CLASS
# -----------------------------

class_survival = (
    df_clean
    .groupby("pclass", observed=True)["survived"]
    .mean()
    .reset_index()
)

class_survival["survival_rate"] = (
    class_survival["survived"] * 100
)

print("\nSurvival Rate by Passenger Class:")
display(class_survival)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=class_survival,
    x="pclass",
    y="survival_rate"
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")

plt.show()


# -----------------------------
# 19. SURVIVAL BY EMBARKATION
# -----------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_clean,
    x="embarked",
    hue="survived"
)

plt.title("Survival by Embarkation Port")
plt.xlabel("Embarkation Port")
plt.ylabel("Number of Passengers")

plt.show()


# -----------------------------
# 20. CORRELATION HEATMAP
# -----------------------------

correlation_data = df_clean[
    ["survived", "age", "sibsp", "parch", "fare"]
].copy()

correlation_matrix = correlation_data.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.show()


# -----------------------------
# 21. AGE VS FARE
# -----------------------------

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x="age",
    y="fare",
    hue="survived",
    alpha=0.7
)

plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()


# ================================================================
# INTERACTIVE VISUALIZATIONS
# ================================================================


# -----------------------------
# 22. INTERACTIVE AGE CHART
# -----------------------------

fig1 = px.histogram(
    df_clean,
    x="age",
    color="survived",
    nbins=30,
    title="Interactive Age Distribution by Survival",
    labels={
        "age": "Age",
        "survived": "Survived"
    }
)

fig1.show()


# -----------------------------
# 23. INTERACTIVE AGE VS FARE
# -----------------------------

fig2 = px.scatter(
    df_clean,
    x="age",
    y="fare",
    color="survived",
    size="fare",
    hover_data=[
        "sex",
        "pclass",
        "embarked"
    ],
    title="Interactive Age vs Fare"
)

fig2.show()


# -----------------------------
# 24. INTERACTIVE CLASS/GENDER
# -----------------------------

dashboard_data = (
    df_clean
    .groupby(
        ["pclass", "sex"],
        observed=True
    )["survived"]
    .mean()
    .reset_index()
)

dashboard_data["survival_rate"] = (
    dashboard_data["survived"] * 100
)

fig3 = px.bar(
    dashboard_data,
    x="pclass",
    y="survival_rate",
    color="sex",
    barmode="group",
    text="survival_rate",
    title="Survival Rate by Class and Gender",
    labels={
        "pclass": "Passenger Class",
        "survival_rate": "Survival Rate (%)",
        "sex": "Gender"
    }
)

fig3.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig3.show()


# ================================================================
# KEY FINDINGS
# ================================================================

print("\n" + "=" * 70)
print("7. KEY FINDINGS")
print("=" * 70)

overall_survival = (
    df_clean["survived"].mean() * 100
)

male_survival = (
    df_clean[
        df_clean["sex"] == "male"
    ]["survived"].mean() * 100
)

female_survival = (
    df_clean[
        df_clean["sex"] == "female"
    ]["survived"].mean() * 100
)

average_age = df_clean["age"].mean()
average_fare = df_clean["fare"].mean()

print(
    f"\n1. Overall survival rate: "
    f"{overall_survival:.2f}%"
)

print(
    f"\n2. Female survival rate: "
    f"{female_survival:.2f}%"
)

print(
    f"   Male survival rate: "
    f"{male_survival:.2f}%"
)

print("\n3. Survival rate by passenger class:")

for passenger_class, rate in class_survival[
    ["pclass", "survival_rate"]
].values:

    print(
        f"   Class {int(passenger_class)}: "
        f"{rate:.2f}%"
    )

print(
    f"\n4. Average passenger age: "
    f"{average_age:.2f} years"
)

print(
    f"\n5. Average passenger fare: "
    f"${average_fare:.2f}"
)


# ================================================================
# DATA QUALITY REPORT
# ================================================================

print("\n" + "=" * 70)
print("8. FINAL DATA QUALITY REPORT")
print("=" * 70)

quality_report = pd.DataFrame({
    "Column": df_clean.columns,
    "Data Type": [
        str(df_clean[col].dtype)
        for col in df_clean.columns
    ],
    "Missing Values": [
        df_clean[col].isnull().sum()
        for col in df_clean.columns
    ],
    "Unique Values": [
        df_clean[col].nunique()
        for col in df_clean.columns
    ]
})

quality_report["Missing %"] = (
    quality_report["Missing Values"] /
    len(df_clean) * 100
).round(2)

display(quality_report)


# ================================================================
# SAVE CLEANED DATASET
# ================================================================

df_clean.to_csv(
    "titanic_cleaned_dataset.csv",
    index=False
)

df_clean.to_excel(
    "titanic_cleaned_dataset.xlsx",
    index=False
)

# Save correlation heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig(
    "titanic_correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# ================================================================
# FINAL SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("                 PROJECT COMPLETED")
print("=" * 70)

print("\nDataset:")
print("Titanic Passenger Dataset")

print(f"\nFinal Rows: {df_clean.shape[0]}")
print(f"Final Columns: {df_clean.shape[1]}")

print("\nTasks Completed:")
print("✅ Dataset loading")
print("✅ Data exploration")
print("✅ Missing value analysis")
print("✅ Missing value handling")
print("✅ Duplicate detection/removal")
print("✅ Column name cleaning")
print("✅ Data type processing")
print("✅ Outlier detection")
print("✅ Outlier treatment")
print("✅ Statistical analysis")
print("✅ Data visualization")
print("✅ Correlation analysis")
print("✅ Interactive visualizations")
print("✅ Key insights")
print("✅ Data quality report")
print("✅ Cleaned CSV export")
print("✅ Excel export")

print("\nGenerated Files:")
print("📁 titanic_cleaned_dataset.csv")
print("📁 titanic_cleaned_dataset.xlsx")
print("📁 titanic_correlation_heatmap.png")

print("\n" + "=" * 70)
print("              ALL TASKS COMPLETED SUCCESSFULLY!")
print("=" * 70)
```
